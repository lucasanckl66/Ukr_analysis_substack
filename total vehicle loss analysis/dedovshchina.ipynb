```python
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle

rows = [
    ["Approximate authorized strength", "130–140", "90–110"],
    ["Tactical enlisted-leader billets", "37–40", "10–13"],
    ["Leaders per 100 personnel", "28–30", "10–13"],
    ["Soldiers per enlisted leader", "3–4", "8–10"]
]

columns = [
    "MEASURE",
    "U.S. ARMY\nRIFLE COMPANY",
    "RUSSIAN MOTOR-RIFLE\nCOMPANY"
]

fig, ax = plt.subplots(figsize=(12, 6.6))
fig.patch.set_facecolor("#f7f5f0")
ax.set_facecolor("#f7f5f0")
ax.axis("off")

fig.add_artist(Rectangle(
    (0.055, 0.925), 0.045, 0.009,
    transform=fig.transFigure,
    color="#c9362b",
    linewidth=0
))

fig.text(
    0.055, 0.86,
    "The Leadership Gap",
    fontfamily="DejaVu Serif",
    fontsize=28,
    fontweight="bold",
    color="#171717",
    ha="left"
)

fig.text(
    0.055, 0.81,
    "U.S. rifle companies place far more enlisted leaders among their ranks.",
    fontfamily="DejaVu Sans",
    fontsize=13.5,
    color="#55514d",
    ha="left"
)

table = ax.table(
    cellText=rows,
    colLabels=columns,
    cellLoc="left",
    colLoc="left",
    colWidths=[0.46, 0.27, 0.27],
    bbox=[0.045, 0.15, 0.91, 0.57]
)

table.auto_set_font_size(False)

for (row, col), cell in table.get_celld().items():
    cell.PAD = 0.045
    cell.set_edgecolor("#cbc7c0")

    if row == 0:
        cell.set_facecolor("#252525")
        cell.set_text_props(
            fontfamily="DejaVu Sans",
            fontsize=10.5,
            fontweight="bold",
            color="white",
            linespacing=1.35
        )
        cell.visible_edges = "closed"
        cell.set_linewidth(0)

    else:
        cell.set_facecolor("#f7f5f0" if row % 2 else "#ece9e3")
        cell.set_text_props(
            fontfamily="DejaVu Sans",
            fontsize=13,
            color="#262626"
        )
        cell.visible_edges = "B"
        cell.set_linewidth(0.7)

        if col > 0:
            cell.get_text().set_ha("center")
            cell.get_text().set_fontfamily("DejaVu Serif")
            cell.get_text().set_fontsize(15)

            if row in [2, 3]:
                cell.get_text().set_fontweight("bold")

    if col > 0 and row == 0:
        cell.get_text().set_ha("center")

for col in range(3):
    table[(4, col)].visible_edges = ""

table.scale(1, 1.75)

fig.text(
    0.055, 0.085,
    "Note: Approximate authorized structure. “Enlisted leaders” includes tactical "
    "leadership billets below the commissioned-officer level.",
    fontfamily="DejaVu Sans",
    fontsize=9.5,
    color="#68635e",
    ha="left"
)

fig.text(
    0.945, 0.045,
    "SOURCE: marines.mil, armyupress.army.mil",
    fontfamily="DejaVu Sans",
    fontsize=8.5,
    fontweight="bold",
    color="#77716b",
    ha="right"
)

plt.savefig(
    "company_comparison_editorial.png",
    dpi=300,
    bbox_inches="tight",
    facecolor=fig.get_facecolor()
)

plt.savefig(
    "figures/leadership_us_rus.png",
    dpi=300,
    bbox_inches="tight",
    facecolor="white"
)
plt.show()
```
